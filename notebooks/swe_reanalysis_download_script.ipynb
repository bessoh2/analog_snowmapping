### Create and update the download script for Snow Depth files:

```python
import re
import shutil

for year in range(1990, 2021):
    year_str = str(year)
    folder_path = os.path.join(BASE_DIR, year_str)

    if not os.path.isdir(folder_path):
        continue

    # Find original .sh file
    sh_files = [f for f in os.listdir(folder_path) if f.endswith("_edited.sh") and not f.endswith("_sd.sh")]
    if not sh_files:
        continue

    orig_sh_file = sh_files[0]
    orig_sh_path = os.path.join(folder_path, orig_sh_file)

    # Build new filename
    base_name, ext = os.path.splitext(orig_sh_file)
    sd_sh_file = f"{base_name}_sd{ext}"
    sd_sh_path = os.path.join(folder_path, sd_sh_file)

    # Read the original file
    with open(orig_sh_path, "r") as f:
        content = f.read()

    # Find all .nc files with SWE_SCA in them
    nc_files = re.findall(r'(\S*SWE_SCA\S*\.nc)', content)

    # Replace them with SD variants
    for old_nc in nc_files:
        new_nc = old_nc.replace("SWE_SCA", "SD")
        content = content.replace(old_nc, new_nc)
        print(f"[{year}] {old_nc} → {new_nc}")

    # Save to new _sd.sh file
    with open(sd_sh_path, "w") as f:
        f.write(content)

    print(f"[WRITE] Created: {sd_sh_file}")
```
